### The "So What" for Refinery Economics

In a refinery, **fractions are separated by boiling points**. 
* **If the cut point is set too low (overly conservative):** The operator is "giving away margin"—valuable diesel molecules (LCO) spill into cheap, low-value heavy slurry/fuel oil, or high-value gasoline molecules (HN) spill into diesel.
* **If the cut point is set too high (too aggressive):** Heavy, dirty components bleed into the product, causing **off-spec diesel or fouled reformer catalysts**, which can trigger hundreds of thousands of dollars in penalties or reprocessing costs.

Because lab tests only return every 8 hours, operators deliberately run conservative cut points, **giving away $0.5M–$1M+ per year in margin** just to stay safe.

### 2. What is a "Cut Point"?

The **Cut Point** is the operational boundary line between two products.
* Think of the fractionator like a vertical sieve: where you set the tray draw rate and temperature determines where **Heavy Naphtha ends and Light Cycle Oil begins** (the HN/LCO cut point), or where **LCO ends and Slurry begins** (the LCO/Slurry cut point).
* Operators adjust this using target set points (in this project: `SP_HN_T98` and `SP_LCO_T98`) which regulate draw valve positions and reflux pumparounds.

### 3. What is T98?

**T98** is a standard laboratory distillation metric (derived from ASTM distillation tests):
* **Definition:** It is the temperature at which **98% of the liquid sample has boiled off** into vapor (representing the "heavy tail" or heaviest molecules in that batch).
* **Why Refineries Track It:** 
  * Specifications for diesel and naphtha have strict limits on their heavy end. 
  * If the **LCO T98 is too high** (e.g. above $760^\circ\text{F}$), heavy polyaromatics and slurry slip into the diesel, causing it to fail ASTM diesel distillation, flash-point, or cetane specifications.
  * If the **HN T98 is too high**, heavy aromatics end up in the reformer feed, which rapidly cokes up and poisons expensive platinum catalysts.

### 2. What Is the Heavy Oil Called?

The actual FCC feed is typically a blend of:
1. **VGO (Vacuum Gas Oil) / HVGO (Heavy Vacuum Gas Oil):** The primary and most common FCC feedstock. Heavy distillate boiling between ~650 °F and 1050 °F.

### 3. Why Do Refinery Engineers Say "Crude Change" or "Crude Slate"?

Even though raw crude doesn't enter the FCC, **the VGO inherits the chemical DNA of the crude oil processed upstream**:

* **If the refinery runs Arab Heavy / Maya crude:** The resulting VGO is dense, highly aromatic, and packed with Conradson Carbon (CCR) and heavy metals (Nickel, Vanadium). When it hits the FCC, it produces massive coke on the catalyst and threatens to overheat the regenerator.
* **If the refinery runs Bonny Light / Brent crude:** The resulting VGO is light and paraffinic. It cracks easily into high-yield gasoline and light olefins with very low coke make.

Because crude oil is purchased and blended upstream in cargo batches (switching every 12–48 hours), operators refer to the incoming transition as the **"Crude Switch"** or **"Feed Slate Transition."**
